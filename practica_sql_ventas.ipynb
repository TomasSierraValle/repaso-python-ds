```python
# Importamos SQLite para trabajar con SQL desde Python
import sqlite3
import pandas as pd

# Creamos una base de datos en memoria
conexion = sqlite3.connect(":memory:")

# Creamos la tabla de ventas
conexion.execute("""
CREATE TABLE ventas_tecnologia (
    id_venta INTEGER PRIMARY KEY,
    producto TEXT,
    categoria TEXT,
    precio_unitario REAL,
    cantidad INTEGER,
    fecha TEXT,
    pais TEXT
)
""")

# Insertamos datos ficticios para practicar las consultas
datos = [
    (1, "Laptop", "Computadoras", 1200, 5, "2026-01-10", "Colombia"),
    (2, "Mouse", "Accesorios", 25, 20, "2026-01-12", "Argentina"),
    (3, "Monitor", "Computadoras", 350, 10, "2026-01-15", "Colombia"),
    (4, "Teclado", "Accesorios", 80, 15, "2026-01-18", "Chile"),
    (5, "Tablet", "Dispositivos móviles", 650, 8, "2026-01-20", "Colombia"),
    (6, "Auriculares", "Accesorios", 150, 12, "2026-01-22", "Argentina"),
    (7, "Smartphone", "Dispositivos móviles", 900, 10, "2026-01-25", "Colombia"),
    (8, "Webcam", None, 120, 5, "2026-01-27", "Chile"),
    (9, "Laptop", "Computadoras", 1100, 4, "2026-01-29", "Argentina"),
    (10, "Monitor", "Computadoras", 550, 6, "2026-02-01", "Colombia")
]

conexion.executemany("""
INSERT INTO ventas_tecnologia
(id_venta, producto, categoria, precio_unitario, cantidad, fecha, pais)
VALUES (?, ?, ?, ?, ?, ?, ?)
""", datos)

conexion.commit()

# Verificamos que los datos se hayan cargado correctamente
df = pd.read_sql_query(
    "SELECT * FROM ventas_tecnologia",
    conexion
)

display(df)
```

```python
# Obtenemos los productos y sus precios ordenados por nombre
consulta = """
SELECT producto, precio_unitario
FROM ventas_tecnologia
ORDER BY producto ASC;
"""

display(pd.read_sql_query(consulta, conexion))
```

```python
# Filtramos las ventas de Colombia con precio superior a 500
consulta = """
SELECT *
FROM ventas_tecnologia
WHERE pais = 'Colombia'
  AND precio_unitario > 500;
"""

display(pd.read_sql_query(consulta, conexion))
```

```python
# Identificamos registros sin categoría
consulta = """
SELECT *
FROM ventas_tecnologia
WHERE categoria IS NULL;
"""

display(pd.read_sql_query(consulta, conexion))
```

```python
# Calculamos los ingresos totales por categoría
consulta = """
SELECT
    categoria,
    SUM(cantidad * precio_unitario) AS ingresos_totales
FROM ventas_tecnologia
GROUP BY categoria;
"""

display(pd.read_sql_query(consulta, conexion))
```

```python
# Mostramos únicamente las categorías con ingresos superiores a 10000
consulta = """
SELECT
    categoria,
    SUM(cantidad * precio_unitario) AS ingresos_totales
FROM ventas_tecnologia
GROUP BY categoria
HAVING SUM(cantidad * precio_unitario) > 10000;
"""

display(pd.read_sql_query(consulta, conexion))
```
