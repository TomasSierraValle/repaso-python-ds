```python
# DEFINICIÓN DE VARIABLES


# Definimos los datos de un producto
nombre_producto = "Teclado"
precio_producto = 85.50
stock_producto = 20
tiene_descuento = True

# Mostramos un resumen del producto utilizando f-strings
print(f"Producto: {nombre_producto}")
print(f"Precio: ${precio_producto}")
print(f"Stock disponible: {stock_producto} unidades")
print(f"¿Tiene descuento?: {tiene_descuento}")


# LÓGICA DE NEGOCIO

# Lista de precios de distintos productos
precios = [45.50, 120, 85, 250.75, 100]

# Recorremos la lista y clasificamos cada precio
for precio in precios:
    if precio > 100:
        print(f"Precio ${precio}: Caro")
    else:
        print(f"Precio ${precio}: Económico")


# DICCIONARIO DE PRODUCTOS


# Diccionario con productos y sus cantidades en stock
almacen = {
    "Teclado": 20,
    "Mouse": 35,
    "Monitor": 10
}

# Agregamos un nuevo producto
almacen["Auriculares"] = 15

# Consultamos e imprimimos el stock de un producto
print(f"Stock de Mouse: {almacen['Mouse']} unidades")

# Mostramos el inventario completo
print(almacen)



# FUNCIÓN ESTADÍSTICA


# Calculamos la suma, el promedio y la cantidad de números
def resumen_estadistico(numeros):
    suma_total = sum(numeros)
    promedio = suma_total / len(numeros)
    cantidad_elementos = len(numeros)

    return {
        "suma_total": suma_total,
        "promedio": promedio,
        "cantidad_elementos": cantidad_elementos
    }

# Llamamos a la función con una lista de ejemplo
resultado = resumen_estadistico([10, 20, 30])

# Imprimimos el resultado
print(resultado)
```
